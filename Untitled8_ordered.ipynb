# 4. Exploratory Data Analysis (EDA) , Calculating Technical Indicators & Trend analysing

```python
stock[['Open', 'High', 'Low', 'Close', 'Volume']].describe()
```

```python
stock['Daily_Return']= stock['Close'].pct_change()*100
stock[["Date", "Close", "Daily_Return"]].head()
```

```python
stock['EMA_20']=stock['Close'].ewm(span=20, adjust=False).mean()
```

```python
stock['EMA_50']=stock['Close'].ewm(span=50, adjust=False).mean()
stock[['Close','EMA_20','EMA_50']].tail()
```

**Understand the trading interpretation**

**Potential bullish signal**

EMA 20 crosses above EMA 50. This can indicate improving momentum.







**Potential bearish signal**

EMA 20 crosses below EMA 50. This can indicate weakening momentum.



These are signals, not guaranteed predictions. Moving averages lag behind price, so crossovers can occur after a move has already started.

```python
stock['Trend'] = np.where(stock['EMA_20'] > stock['EMA_50'],'Bullish','Bearish')
stock[['Date', 'Close', 'EMA_20', 'EMA_50', 'Trend']].tail()
```

**Create a crossover signal**
We'll use a new column called Signal.

**The logic is:**

**Bearish crossover:** EMA 20 was above EMA 50 in the previous row, but is now below it.

**Bullish crossover:** EMA 20 was below EMA 50 in the previous row, but is now above it.

```python
stock['Previous_Trend'] = stock['Trend'].shift(1)
stock[['Date', 'Close', 'EMA_20', 'EMA_50', 'Trend', 'Previous_Trend']].tail()
```

```python
stock['signal'] = np.where((stock['Trend'] == 'Bullish') & (stock['Previous_Trend'] == 'Bearish'), 'Bullish Crossover',
                   np.where((stock['Trend'] == 'Bearish') & (stock['Previous_Trend'] == 'Bullish'), 'Bearish Crossover', 'No Crossover'))
stock[['Date', 'Close', 'EMA_20', 'EMA_50', 'Trend', 'Previous_Trend', 'signal']].tail()
```

```python
stock[stock['signal'] != 'No Crossover'][['Date', 'Close', 'EMA_20', 'EMA_50', 'signal']]
```

```python
stock['signal'].value_counts()
```

MACD uses two exponential moving averages:
- **Fast EMA:** 12 periods
- **Slow EMA: 26** periods
- **MACD line:** Fast EMA minus slow EMA

```python
stock['EMA_12'] = stock['Close'].ewm(span=12, adjust=False).mean()
stock['EMA_26'] = stock['Close'].ewm(span=26, adjust=False).mean()
stock['MACD'] = stock['EMA_12'] - stock['EMA_26']
```

```python
stock['MACD_Signal'] = stock['MACD'].ewm(span=9, adjust=False).mean()
stock[['Date', 'Close', 'MACD', 'MACD_Signal']].tail()
```

Positive histogram: MACD is above its signal line.

Negative histogram:MACD is below its signal line.

Increasing negative magnitude: The gap below the signal line is widening, suggesting strengthening negative momentum during these observations.

```python
stock['MACD_Histogram'] = stock['MACD'] - stock['MACD_Signal']
stock[['Date', 'Close', 'MACD', 'MACD_Signal', 'MACD_Histogram']].tail()
```

create a new column called MACD_Signal_Type.

The logic is:
- Bullish crossover: MACD moves from below the signal line to above it.
- Bearish crossover: MACD moves from above the signal line to below it.
- No Crossover: Neither condition occurs.

```python
stock['Previous_MACD_Histogram'] = stock['MACD_Histogram'].shift(1)
stock[['Date', 'Close', 'MACD', 'MACD_Signal', 'MACD_Histogram', 'Previous_MACD_Histogram']].tail()
```

**Bullish crossover:** Previous histogram was below zero, and current histogram is above zero.



**Bearish crossover:** Previous histogram was above zero, and current histogram is below zero.



**No crossover:** Otherwise.

```python
stock['MACD_Signal_Type']= np.where((stock['MACD_Histogram'] > 0) & (stock['Previous_MACD_Histogram'] < 0), 'Bullish Crossover',
                                     np.where((stock['MACD_Histogram'] < 0) & (stock['Previous_MACD_Histogram'] > 0), 'Bearish Crossover', 'No Crossover'))
stock[['Date', 'Close', 'MACD', 'MACD_Signal', 'MACD_Histogram', 'Previous_MACD_Histogram', 'MACD_Signal_Type']].tail()
```

ADX measures how strong a trend is, regardless of whether the price is moving up or down.

```python
!pip install ta
```

```python
from ta.trend import ADXIndicator
```

```python
adx_indicator = ADXIndicator(
    high=stock['High'],
    low=stock['Low'],
    close=stock['Close'],
    window=14
)

stock['ADX'] = adx_indicator.adx()
stock['DI_Positive'] = adx_indicator.adx_pos()
stock['DI_Negative'] = adx_indicator.adx_neg()
```

Understand the three columns:
- **ADX:** Measures trend strength.
- **DI_Positive:** Measures positive directional movement.
- **DI_Negative:** Measures negative directional movement.

```python
stock[['Date', 'Close', 'ADX',
       'DI_Positive', 'DI_Negative']].tail(10)
```

Common ADX guidelines:
- Below 20: Often a weak or sideways trend.
- 20–25: A trend may be developing.
- Above 25: Often indicates a stronger trend.

These are guidelines, not guaranteed trading signals.

Volume Analysis

Volume tells us how many shares were traded during a particular day.
- High volume: Many shares were traded.
- Low volume: Fewer shares were traded.
- Price rising + high volume: The upward move may have stronger participation.
- Price falling + high volume: The downward move may have stronger participation.

```python
stock.loc[stock['Volume'].idxmax()]
```

```python

average_volume = stock['Volume'].mean()
stock['High_Volume'] = stock['Volume'] > (2 * average_volume)
print("Average Volume:", average_volume)
print("\nHigh-volume days:")
print(stock.loc[stock['High_Volume'], ['Date', 'Close', 'Volume']].tail(10))
```

Classify the stock trend as Strong Bullish, Strong Bearish, or Mixed/Weak using EMA, MACD, ADX, and directional indicators.

```python

stock['Overall_Trend'] = np.where(
    (stock['EMA_20'] > stock['EMA_50']) &
    (stock['MACD'] > stock['MACD_Signal']) &
    (stock['ADX'] > 25) &
    (stock['DI_Positive'] > stock['DI_Negative']),
    'Strong Bullish',
    np.where(
        (stock['EMA_20'] < stock['EMA_50']) &
        (stock['MACD'] < stock['MACD_Signal']) &
        (stock['ADX'] > 25) &
        (stock['DI_Negative'] > stock['DI_Positive']),
        'Strong Bearish',
        'Mixed/Weak'
    )
)

print(stock['Overall_Trend'].value_counts())
```

ADX measures strength, not direction. The +DI and −DI lines help assess direction.
